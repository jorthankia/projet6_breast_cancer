```python


print("\n" + "="*60)
print("📊 CHARGEMENT DES DONNÉES")
print("="*60)

# Chemins des fichiers dans Google Drive
clinical_file = "/content/drive/MyDrive/brca_tcga_data_clinical_patient.txt"
mrna_file = "/content/drive/MyDrive/brca_tcga_data_RNA_Seq_v2_expression_median.txt"
pam50_file = "/content/drive/MyDrive/PAM50_genes.txt"


# Chargement des données cliniques
clinical_df = pd.read_csv(clinical_file, sep='\t', skiprows=4)
print(f"✅ Données cliniques chargées : {clinical_df.shape[0]} patients, {clinical_df.shape[1]} colonnes")

# Chargement des données d'expression
mrna_df = pd.read_csv(mrna_file, sep='\t')
print(f"✅ Données d'expression chargées : {mrna_df.shape[0]} gènes, {mrna_df.shape[1]} échantillons")

# Chargement des gènes PAM50
pam50_df = pd.read_csv(pam50_file, sep='\t')
pam50_genes = pam50_df['Hugo_Symbol'].values
print(f"✅ {len(pam50_genes)} gènes PAM50 chargés")

# ============================================
# ÉTAPE 3 : NETTOYAGE ET PRÉPARATION DES DONNÉES
# ============================================

print("\n" + "="*60)
print("🧹 NETTOYAGE ET PRÉPARATION DES DONNÉES")
print("="*60)

# Nettoyage des ID patients
clinical_df['PATIENT_ID'] = clinical_df['PATIENT_ID'].str.replace('_', '-', regex=False)

# Suppression de la colonne Entrez_Gene_Id si présente
if 'Entrez_Gene_Id' in mrna_df.columns:
    mrna_df.drop(columns=['Entrez_Gene_Id'], inplace=True)
    print("✅ Colonne Entrez_Gene_Id supprimée")

# Filtrage des gènes PAM50
mrna_df = mrna_df[mrna_df['Hugo_Symbol'].isin(pam50_genes)]
print(f"✅ Données d'expression filtrées : {mrna_df.shape[0]} gènes PAM50")

# Transformation des données d'expression (patients en lignes, gènes en colonnes)
mrna_long = mrna_df.set_index('Hugo_Symbol').T.reset_index()
mrna_long = mrna_long.rename(columns={'index': 'PATIENT_ID'})
mrna_long['PATIENT_ID'] = mrna_long['PATIENT_ID'].str.replace('_', '-', regex=False).str[:12]

# Fusion des données cliniques et d'expression
clinical_subset = clinical_df[['PATIENT_ID', 'SUBTYPE']]
data = pd.merge(clinical_subset, mrna_long, on='PATIENT_ID', how='inner')

# Nettoyage des sous-types
data = data[data['SUBTYPE'] != 'NA']
data['SUBTYPE'] = data['SUBTYPE'].str.replace('_', '-', regex=False)
data['SUBTYPE'] = data['SUBTYPE'].astype('category')

print(f"✅ Données fusionnées : {data.shape[0]} patients, {data.shape[1]} colonnes")
print(f"\n📊 Distribution des sous-types :")
print(data['SUBTYPE'].value_counts())

# ============================================
# ÉTAPE 4 : SÉLECTION DES SOUS-TYPES ET TRANSFORMATION LOG2
# ============================================

print("\n" + "="*60)
print("🎯 SÉLECTION DES SOUS-TYPES ET TRANSFORMATION LOG2")
print("="*60)

# Définition des sous-types pour l'analyse
# Tu peux changer ces valeurs selon tes besoins
CONSTANT_SUBTYPE = 'BRCA-Normal'  # Sous-type constant
COMPARATOR1 = 'BRCA-LumA'         # Premier comparateur
COMPARATOR2 = 'BRCA-Her2'         # Deuxième comparateur

print(f"🎯 Sous-type constant : {CONSTANT_SUBTYPE}")
print(f"📊 Comparaison 1 : {CONSTANT_SUBTYPE} vs {COMPARATOR1}")
print(f"📊 Comparaison 2 : {CONSTANT_SUBTYPE} vs {COMPARATOR2}")

# Filtrage des sous-types
subtypes_to_keep = [CONSTANT_SUBTYPE, COMPARATOR1, COMPARATOR2]
df_filtered = data[data['SUBTYPE'].isin(subtypes_to_keep)].copy()
print(f"\n✅ Données filtrées : {df_filtered.shape[0]} patients")

# Transformation log2
gene_cols = df_filtered.columns.drop(['PATIENT_ID', 'SUBTYPE'])
df_transformed = df_filtered.copy()
df_transformed[gene_cols] = df_transformed[gene_cols].apply(lambda x: np.log2(x + 1))

print(f"✅ Transformation log2 appliquée")

# ============================================
# ÉTAPE 5 : ANALYSE D'EXPRESSION DIFFÉRENTIELLE
# ============================================

print("\n" + "="*60)
print("🔬 ANALYSE D'EXPRESSION DIFFÉRENTIELLE")
print("="*60)

def differential_expression(df, subtype_constant, target_subtype):
    """
    Effectue l'analyse d'expression différentielle entre deux sous-types
    """
    # Séparation des échantillons
    df_constant = df[df['SUBTYPE'] == subtype_constant]
    df_target = df[df['SUBTYPE'] == target_subtype]

    print(f"\n📊 {subtype_constant} vs {target_subtype}:")
    print(f"   - {subtype_constant}: {len(df_constant)} échantillons")
    print(f"   - {target_subtype}: {len(df_target)} échantillons")

    # Identification des gènes
    gene_cols = df.columns.drop(['PATIENT_ID', 'SUBTYPE'])
    results = []

    for gene in gene_cols:
        expr_constant = df_constant[gene].values
        expr_target = df_target[gene].values

        # Test t de Student
        t_stat, p_val = ttest_ind(expr_constant, expr_target, equal_var=False)

        # Log2 Fold Change
        mean_constant = np.mean(expr_constant)
        mean_target = np.mean(expr_target)
        log2fc = mean_constant - mean_target

        results.append({
            'gene': gene,
            'log2FC': log2fc,
            'p_value': p_val,
            'mean_constant': mean_constant,
            'mean_target': mean_target
        })

    # Création du DataFrame des résultats
    de_results = pd.DataFrame(results)

    # Correction FDR
    de_results['p_adjusted'] = multipletests(de_results['p_value'], method='fdr_bh')[1]

    # Classification des gènes
    de_results['regulation'] = 'Non significatif'
    de_results.loc[(de_results['log2FC'] > 0.5) & (de_results['p_adjusted'] < 0.05), 'regulation'] = f'Surexprimé dans {subtype_constant}'
    de_results.loc[(de_results['log2FC'] < -0.5) & (de_results['p_adjusted'] < 0.05), 'regulation'] = f'Surexprimé dans {target_subtype}'

    # Tri
    de_results = de_results.sort_values('log2FC', ascending=False)

    # Statistiques
    n_up = sum(de_results['regulation'] == f'Surexprimé dans {subtype_constant}')
    n_down = sum(de_results['regulation'] == f'Surexprimé dans {target_subtype}')
    print(f"   📈 Gènes surexprimés dans {subtype_constant}: {n_up}")
    print(f"   📉 Gènes surexprimés dans {target_subtype}: {n_down}")

    return de_results

# Exécution des deux analyses
de_results_1 = differential_expression(df_transformed, CONSTANT_SUBTYPE, COMPARATOR1)
de_results_2 = differential_expression(df_transformed, CONSTANT_SUBTYPE, COMPARATOR2)

# Sauvegarde des résultats
de_results_1.to_csv(f'DE_{CONSTANT_SUBTYPE}_vs_{COMPARATOR1}.csv', index=False)
de_results_2.to_csv(f'DE_{CONSTANT_SUBTYPE}_vs_{COMPARATOR2}.csv', index=False)
print("\n💾 Résultats DE sauvegardés en CSV")

# ============================================
# ÉTAPE 6 : VOLCAN PLOTS
# ============================================

print("\n" + "="*60)
print("📊 CRÉATION DES VOLCAN PLOTS")
print("="*60)

def plot_volcano(de_results, subtype_constant, target_subtype):
    """
    Crée un volcan plot
    """
    fig, ax = plt.subplots(figsize=(12, 8))

    colors = {
        f'Surexprimé dans {subtype_constant}': '#FF4444',
        f'Surexprimé dans {target_subtype}': '#4444FF',
        'Non significatif': '#CCCCCC'
    }

    for reg, color in colors.items():
        subset = de_results[de_results['regulation'] == reg]
        if len(subset) > 0:
            ax.scatter(subset['log2FC'],
                      -np.log10(subset['p_adjusted'] + 1e-300),
                      c=color, alpha=0.6, s=30, label=reg)

    # Lignes de seuil
    ax.axhline(y=-np.log10(0.05), color='black', linestyle='--', alpha=0.5)
    ax.axvline(x=0.5, color='black', linestyle='--', alpha=0.5)
    ax.axvline(x=-0.5, color='black', linestyle='--', alpha=0.5)

    ax.set_xlabel('log2 Fold Change', fontsize=14)
    ax.set_ylabel('-log10(Adjusted p-value)', fontsize=14)
    ax.set_title(f'Volcan Plot : {subtype_constant} vs {target_subtype}', fontsize=16)
    ax.legend(loc='upper left', fontsize=12)
    ax.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.savefig(f'volcano_{subtype_constant}_vs_{target_subtype}.png', dpi=300, bbox_inches='tight')
    plt.show()
    print(f"✅ Volcan plot sauvegardé : volcano_{subtype_constant}_vs_{target_subtype}.png")
    return fig

# Création des volcan plots
plot_volcano(de_results_1, CONSTANT_SUBTYPE, COMPARATOR1)
plot_volcano(de_results_2, CONSTANT_SUBTYPE, COMPARATOR2)

# ============================================
# ÉTAPE 7 : EXTRACTION DES GÈNES SUREXPRIMÉS
# ============================================

print("\n" + "="*60)
print("🧬 EXTRACTION DES GÈNES SUREXPRIMÉS")
print("="*60)

def get_upregulated_genes(de_results, subtype_constant):
    """Extrait les gènes surexprimés dans le sous-type constant"""
    upregulated = de_results[de_results['regulation'] == f'Surexprimé dans {subtype_constant}']
    genes = upregulated['gene'].tolist()
    fc = upregulated['log2FC'].tolist()
    pvals = upregulated['p_adjusted'].tolist()
    return genes, fc, pvals

# Extraction
genes_up_1, fc_1, pvals_1 = get_upregulated_genes(de_results_1, CONSTANT_SUBTYPE)
genes_up_2, fc_2, pvals_2 = get_upregulated_genes(de_results_2, CONSTANT_SUBTYPE)

print(f"\n✅ {CONSTANT_SUBTYPE} vs {COMPARATOR1}: {len(genes_up_1)} gènes surexprimés")
print(f"✅ {CONSTANT_SUBTYPE} vs {COMPARATOR2}: {len(genes_up_2)} gènes surexprimés")

# Affichage des top gènes
if len(genes_up_1) > 0:
    print(f"\n🏆 Top 5 gènes surexprimés dans {CONSTANT_SUBTYPE} vs {COMPARATOR1}:")
    for i in range(min(5, len(genes_up_1))):
        print(f"   {i+1}. {genes_up_1[i]} (log2FC = {fc_1[i]:.2f}, p = {pvals_1[i]:.2e})")

if len(genes_up_2) > 0:
    print(f"\n🏆 Top 5 gènes surexprimés dans {CONSTANT_SUBTYPE} vs {COMPARATOR2}:")
    for i in range(min(5, len(genes_up_2))):
        print(f"   {i+1}. {genes_up_2[i]} (log2FC = {fc_2[i]:.2f}, p = {pvals_2[i]:.2e})")



# ============================================
# ÉTAPE 8 : ANALYSE ENRICHISSEMENT AVEC ENRICHR
# MÉTHODE ALTERNATIVE - TÉLÉCHARGEMENT DIRECT
# ============================================

print("\n" + "="*60)
print("🔬 ANALYSE D'ENRICHISSEMENT AVEC ENRICHR")
print("📥 TÉLÉCHARGEMENT DIRECT DES RÉSULTATS")
print("="*60)

import os
import requests
import time

# Création du dossier pour les résultats
results_dir = '/content/enrichr_results'
if not os.path.exists(results_dir):
    os.makedirs(results_dir)
    print(f"✅ Dossier créé : {results_dir}")

# URL de l'API Enrichr
ENRICHR_URL_ADDLIST = 'https://maayanlab.cloud/Enrichr/addList'
ENRICHR_URL_EXPORT = 'https://maayanlab.cloud/Enrichr/export'

# Définition des bases de données à interroger
# Sélection des plus pertinentes pour le cancer du sein
DATABASES = [
    'KEGG_2021_Human',
    'GO_Biological_Process_2021',
    'Reactome_2022',
    'GO_Molecular_Function_2021',
    'MSigDB_Hallmark_2020',
    'WikiPathways_2019_Human'
]

print(f"\n📚 Bases de données sélectionnées :")
for db in DATABASES:
    print(f"   - {db}")

# ============================================
# FONCTION POUR ENVOYER UNE LISTE DE GÈNES À ENRICHR
# ============================================

def submit_genes_to_enrichr(genes, description="GeneList"):
    """
    Soumet une liste de gènes à Enrichr et retourne l'ID de la liste
    """
    if not genes:
        print("⚠️ Liste de gènes vide")
        return None

    # Formatage de la liste
    genes_str = '\n'.join(genes)

    # Envoi de la requête
    payload = {
        'list': (None, genes_str),
        'description': (None, description)
    }

    try:
        response = requests.post(ENRICHR_URL_ADDLIST, files=payload, timeout=30)
        if response.ok:
            data = response.json()
            user_list_id = data['userListId']
            print(f"   ✅ Liste soumise avec succès (ID: {user_list_id})")
            return user_list_id
        else:
            print(f"   ❌ Erreur: {response.status_code}")
            return None
    except Exception as e:
        print(f"   ❌ Exception: {str(e)}")
        return None

# ============================================
# FONCTION POUR TÉLÉCHARGER LES RÉSULTATS
# ============================================

def download_enrichr_results(user_list_id, database, output_name, results_dir):
    """
    Télécharge les résultats d'enrichissement pour une base de données
    """
    filename = f"{output_name}_{database}.txt"
    file_path = os.path.join(results_dir, filename)

    export_url = f"{ENRICHR_URL_EXPORT}?userListId={user_list_id}&filename={filename}&backgroundType={database}"

    try:
        # Téléchargement
        response = requests.get(export_url, stream=True, timeout=60)
        response.raise_for_status()

        # Sauvegarde
        with open(file_path, 'wb') as f:
            for chunk in response.iter_content(chunk_size=8192):
                f.write(chunk)

        print(f"   ✅ {database} → {file_path}")
        return file_path
    except Exception as e:
        print(f"   ❌ Erreur pour {database}: {str(e)}")
        return None

# ============================================
# ANALYSE POUR LA COMPARAISON 1
# ============================================

print(f"\n" + "="*60)
print(f"📊 Enrichissement 1: {CONSTANT_SUBTYPE} vs {COMPARATOR1}")
print("="*60)

# Limitation du nombre de gènes pour éviter les erreurs
genes_to_analyze_1 = genes_up_1[:500] if len(genes_up_1) > 500 else genes_up_1
print(f"🧬 {len(genes_to_analyze_1)} gènes analysés")

# Soumission de la liste
user_list_id_1 = submit_genes_to_enrichr(
    genes_to_analyze_1,
    f"{CONSTANT_SUBTYPE}_vs_{COMPARATOR1}"
)

# Téléchargement des résultats
if user_list_id_1:
    print(f"\n📥 Téléchargement des résultats...")
    for database in DATABASES:
        download_enrichr_results(
            user_list_id_1,
            database,
            f"{CONSTANT_SUBTYPE}_vs_{COMPARATOR1}",
            results_dir
        )
        time.sleep(1)  # Pause pour éviter le rate limiting
else:
    print("❌ Échec de la soumission de la liste")

# ============================================
# ANALYSE POUR LA COMPARAISON 2
# ============================================

print(f"\n" + "="*60)
print(f"📊 Enrichissement 2: {CONSTANT_SUBTYPE} vs {COMPARATOR2}")
print("="*60)

# Limitation du nombre de gènes
genes_to_analyze_2 = genes_up_2[:500] if len(genes_up_2) > 500 else genes_up_2
print(f"🧬 {len(genes_to_analyze_2)} gènes analysés")

# Soumission de la liste
user_list_id_2 = submit_genes_to_enrichr(
    genes_to_analyze_2,
    f"{CONSTANT_SUBTYPE}_vs_{COMPARATOR2}"
)

# Téléchargement des résultats
if user_list_id_2:
    print(f"\n📥 Téléchargement des résultats...")
    for database in DATABASES:
        download_enrichr_results(
            user_list_id_2,
            database,
            f"{CONSTANT_SUBTYPE}_vs_{COMPARATOR2}",
            results_dir
        )
        time.sleep(1)
else:
    print("❌ Échec de la soumission de la liste")

# ============================================
# CHARGEMENT ET ANALYSE DES RÉSULTATS
# ============================================

print("\n" + "="*60)
print("📊 CHARGEMENT ET ANALYSE DES RÉSULTATS")
print("="*60)

def load_enrichr_results(file_path):
    """
    Charge un fichier de résultats Enrichr
    """
    try:
        df = pd.read_csv(file_path, sep='\t')
        return df
    except Exception as e:
        print(f"   ❌ Erreur de chargement: {e}")
        return None

# Chargement des résultats pour KEGG et GO
def load_enrichment_data(comparison_name, results_dir):
    """
    Charge les résultats pour toutes les bases de données
    """
    results = {}

    for database in DATABASES:
        filename = f"{comparison_name}_{database}.txt"
        file_path = os.path.join(results_dir, filename)

        if os.path.exists(file_path):
            df = load_enrichr_results(file_path)
            if df is not None and len(df) > 0:
                # Filtrer les termes significatifs (q < 0.05)
                if 'Adjusted P-value' in df.columns:
                    df_sig = df[df['Adjusted P-value'] < 0.05].sort_values('Adjusted P-value')
                    results[database] = df_sig
                    print(f"   ✅ {database}: {len(df_sig)} termes enrichis (q < 0.05)")
                else:
                    print(f"   ⚠️ {database}: Colonne 'Adjusted P-value' non trouvée")
                    results[database] = df
            else:
                print(f"   ❌ {database}: Fichier vide ou invalide")
        else:
            print(f"   ⚠️ {database}: Fichier non trouvé")

    return results

# Chargement des résultats
print(f"\n📥 Chargement des résultats pour {CONSTANT_SUBTYPE} vs {COMPARATOR1}")
enrich_results_1 = load_enrichment_data(f"{CONSTANT_SUBTYPE}_vs_{COMPARATOR1}", results_dir)

print(f"\n📥 Chargement des résultats pour {CONSTANT_SUBTYPE} vs {COMPARATOR2}")
enrich_results_2 = load_enrichment_data(f"{CONSTANT_SUBTYPE}_vs_{COMPARATOR2}", results_dir)

# ============================================
# VISUALISATION DES RÉSULTATS D'ENRICHISSEMENT
# ============================================

print("\n" + "="*60)
print("📊 VISUALISATION DES RÉSULTATS")
print("="*60)

def plot_enrichment_results_from_df(df, subtype_constant, target_subtype, database, top_n=15):
    """
    Visualise les résultats d'enrichissement à partir d'un DataFrame
    """
    if df is None or len(df) == 0:
        print(f"⚠️ Aucune donnée pour {database}")
        return

    # Sélection des top termes
    df_top = df.head(top_n)

    fig, ax = plt.subplots(figsize=(12, 8))

    # Création du barplot
    y_pos = np.arange(len(df_top))

    if 'Adjusted P-value' in df_top.columns:
        p_values = df_top['Adjusted P-value'].values
        x_vals = -np.log10(p_values + 1e-300)  # Éviter log(0)
    else:
        # Utiliser la première colonne numérique
        numeric_cols = df_top.select_dtypes(include=[np.number]).columns
        if len(numeric_cols) > 0:
            x_vals = df_top[numeric_cols[0]].values
        else:
            print(f"⚠️ Aucune colonne numérique trouvée")
            return

    ax.barh(y_pos, x_vals, color=plt.cm.viridis(np.linspace(0.3, 0.9, len(df_top))))

    # Personnalisation
    ax.set_yticks(y_pos)
    if 'Term' in df_top.columns:
        labels = df_top['Term'].str[:60] + ('...' if df_top['Term'].str.len().max() > 60 else '')
        ax.set_yticklabels(labels, fontsize=10)

    ax.set_xlabel('-log10(p-value ajusté)' if 'Adjusted P-value' in df_top.columns else 'Score', fontsize=14)
    ax.set_title(f'Top {top_n} voies enrichies\n{subtype_constant} vs {target_subtype} ({database})', fontsize=16)
    ax.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.savefig(f'enrichment_{subtype_constant}_vs_{target_subtype}_{database}.png', dpi=300, bbox_inches='tight')
    plt.show()
    print(f"✅ Graphique sauvegardé")

# Visualisation pour KEGG
if 'KEGG_2021_Human' in enrich_results_1 and enrich_results_1['KEGG_2021_Human'] is not None:
    plot_enrichment_results_from_df(
        enrich_results_1['KEGG_2021_Human'],
        CONSTANT_SUBTYPE, COMPARATOR1, 'KEGG_2021_Human'
    )

if 'KEGG_2021_Human' in enrich_results_2 and enrich_results_2['KEGG_2021_Human'] is not None:
    plot_enrichment_results_from_df(
        enrich_results_2['KEGG_2021_Human'],
        CONSTANT_SUBTYPE, COMPARATOR2, 'KEGG_2021_Human'
    )

# Visualisation pour GO Biological Process
if 'GO_Biological_Process_2021' in enrich_results_1 and enrich_results_1['GO_Biological_Process_2021'] is not None:
    plot_enrichment_results_from_df(
        enrich_results_1['GO_Biological_Process_2021'],
        CONSTANT_SUBTYPE, COMPARATOR1, 'GO_Biological_Process_2021'
    )

# ============================================
# COMPARAISON DES RÉSULTATS
# ============================================

print("\n" + "="*60)
print("📊 COMPARAISON DES VOIES ENRICHIES")
print("="*60)

def compare_enrichment_results(results_1, results_2, database='KEGG_2021_Human'):
    """
    Compare les résultats d'enrichissement entre deux comparaisons
    """
    if database not in results_1 or database not in results_2:
        print(f"⚠️ {database} non disponible")
        return None, None, None # Return None for all three values

    df1 = results_1[database]
    df2 = results_2[database]

    if df1 is None or df2 is None or len(df1) == 0 or len(df2) == 0:
        print(f"⚠️ {database}: Données insuffisantes")
        return None, None, None # Return None for all three values

    # Récupération des termes
    terms_1 = set(df1['Term'].tolist()) if 'Term' in df1.columns else set()
    terms_2 = set(df2['Term'].tolist()) if 'Term' in df2.columns else set()

    common = terms_1 & terms_2
    specific_1 = terms_1 - terms_2
    specific_2 = terms_2 - terms_1

    print(f"\n📊 {database}:")
    print(f"   - {CONSTANT_SUBTYPE} vs {COMPARATOR1}: {len(terms_1)} termes")
    print(f"   - {CONSTANT_SUBTYPE} vs {COMPARATOR2}: {len(terms_2)} termes")
    print(f"   - Communs: {len(common)}")
    print(f"   - Spécifiques à {CONSTANT_SUBTYPE} vs {COMPARATOR1}: {len(specific_1)}")
    print(f"   - Spécifiques à {CONSTANT_SUBTYPE} vs {COMPARATOR2}: {len(specific_2)}")

    if len(common) > 0:
        print(f"\n🔗 Termes communs (top 5):")
        for term in list(common)[:5]:
            print(f"   • {term}")

    if len(specific_1) > 0:
        print(f"\n🔴 Spécifiques à {CONSTANT_SUBTYPE} vs {COMPARATOR1} (top 3):")
        for term in list(specific_1)[:3]:
            print(f"   • {term}")

    if len(specific_2) > 0:
        print(f"\n🔵 Spécifiques à {CONSTANT_SUBTYPE} vs {COMPARATOR2} (top 3):")
        for term in list(specific_2)[:3]:
            print(f"   • {term}")

    return common, specific_1, specific_2

# Comparaison pour KEGG
print("\n🔍 Comparaison KEGG:")
common_kegg, spec1_kegg, spec2_kegg = compare_enrichment_results(enrich_results_1, enrich_results_2, 'KEGG_2021_Human')

# Comparaison pour GO
print("\n🔍 Comparaison GO Biological Process:")
common_go, spec1_go, spec2_go = compare_enrichment_results(enrich_results_1, enrich_results_2, 'GO_Biological_Process_2021')

# ============================================
# SAUVEGARDE ET RÉSUMÉ FINAL
# ============================================

print("\n" + "="*60)
print("📋 RÉSUMÉ FINAL")
print("="*60)

print(f"""
✅ ANALYSE D'ENRICHISSEMENT TERMINÉE

📁 Résultats sauvegardés dans : {results_dir}
📊 Bases de données analysées : {len(DATABASES)}

Fichiers générés :
""")

# Liste des fichiers générés
for file in os.listdir(results_dir):
    if file.endswith('.txt'):
        print(f"   • {file}")

print(f"\n💡 Pour utiliser ces résultats :")
print(f"   1. Télécharge les fichiers depuis Colab")
print(f"   2. Importe-les dans Excel ou Python pour analyse")
print(f"   3. Utilise les visualisations ci-dessus pour ta présentation")

# Sauvegarde des DataFrames en CSV pour une utilisation future
if enrich_results_1 and 'KEGG_2021_Human' in enrich_results_1:
    if enrich_results_1['KEGG_2021_Human'] is not None:
        enrich_results_1['KEGG_2021_Human'].to_csv(
            f'kegg_results_{CONSTANT_SUBTYPE}_vs_{COMPARATOR1}.csv',
            index=False
        )
        print(f"💾 KEGG résultats sauvegardés en CSV")

if enrich_results_2 and 'KEGG_2021_Human' in enrich_results_2:
    if enrich_results_2['KEGG_2021_Human'] is not None:
        enrich_results_2['KEGG_2021_Human'].to_csv(
            f'kegg_results_{CONSTANT_SUBTYPE}_vs_{COMPARATOR2}.csv',
            index=False
        )
        print(f"💾 KEGG résultats sauvegardés en CSV")

print("\n" + "="*60)
print("✅ ANALYSE COMPLÈTE TERMINÉE !")
print("="*60)

# Construction des réseaux
# network_kegg = build_pathway_network(enrich_results_1, enrich_results_2, 'KEGG_2021_Human') # Commented out as build_pathway_network is not defined
# network_go = build_pathway_network(enrich_results_1, enrich_results_2, 'GO_Biological_Process_2021') # Commented out as build_pathway_network is not defined

# ============================================
# ÉTAPE 9 : HEATMAP DES GÈNES DIFFÉRENTIELLEMENT EXPRIMÉS
# ============================================

print("\n" + "="*60)
print("🔥 HEATMAP DES GÈNES DIFFÉRENTIELLEMENT EXPRIMÉS")
print("="*60)

def plot_heatmap(df, de_results_1, de_results_2, subtype_constant, comparator1, comparator2, top_n=30):
    """Crée un heatmap des gènes différentiellement exprimés"""

    # Sélection des gènes
    genes_1 = de_results_1.nsmallest(top_n//2, 'p_adjusted')['gene'].tolist()
    genes_2 = de_results_2.nsmallest(top_n//2, 'p_adjusted')['gene'].tolist()
    top_genes = list(set(genes_1 + genes_2))[:top_n]

    if len(top_genes) == 0:
        print("⚠️ Aucun gène significatif")
        return None

    # Préparation des données
    df_heatmap = df[df['SUBTYPE'].isin([subtype_constant, comparator1, comparator2])]
    data_heat = df_heatmap[top_genes + ['SUBTYPE']].set_index('SUBTYPE')[top_genes].T

    # Standardisation
    data_heat_std = data_heat.apply(lambda x: (x - x.mean()) / x.std(), axis=1)

    # Palette de couleurs
    subtype_palette = {
        subtype_constant: '#FF4444',
        comparator1: '#4444FF',
        comparator2: '#44BB44'
    }

    # Création du heatmap
    col_colors = df_heatmap['SUBTYPE'].map(subtype_palette)

    plt.figure(figsize=(16, 12))
    g = sns.clustermap(data_heat_std,
                       cmap='RdBu_r',
                       center=0,
                       col_colors=col_colors,
                       figsize=(16, 12),
                       dendrogram_ratio=0.1,
                       cbar_pos=(0.02, 0.8, 0.03, 0.18),
                       row_cluster=True,
                       col_cluster=True,
                       xticklabels=False)

    plt.suptitle(f'Heatmap des top {len(top_genes)} gènes différentiellement exprimés\n'
                 f'Rouge = {subtype_constant} | Bleu = {comparator1} | Vert = {comparator2}',
                 fontsize=14, y=1.02)

    plt.tight_layout()
    plt.savefig('heatmap_top_genes.png', dpi=300, bbox_inches='tight')
    plt.show()
    print("✅ Heatmap sauvegardé : heatmap_top_genes.png")
    return g

# Création du heatmap
plot_heatmap(df_transformed, de_results_1, de_results_2,
             CONSTANT_SUBTYPE, COMPARATOR1, COMPARATOR2)

# ============================================
# ÉTAPE 10 : RÉSUMÉ FINAL
# ============================================

print("\n" + "="*60)
print("📋 RÉSUMÉ FINAL DES RÉSULTATS")
print("="*60)

print(f"\n🎯 Sous-type analysé : {CONSTANT_SUBTYPE}")
print(f"   - Comparaison 1 : {CONSTANT_SUBTYPE} vs {COMPARATOR1}")
print(f"   - Comparaison 2 : {CONSTANT_SUBTYPE} vs {COMPARATOR2}")

print(f"\n📊 Résultats d'expression différentielle :")
print(f"   - {CONSTANT_SUBTYPE} vs {COMPARATOR1}: {len(genes_up_1)} gènes surexprimés")
print(f"   - {CONSTANT_SUBTYPE} vs {COMPARATOR2}: {len(genes_up_2)} gènes surexprimés")

print(f"\n🔬 Interprétation biologique :")
if common_kegg and len(common_kegg) > 0:
    print(f"   ✅ Voies communes (KEGG) caractéristiques du sous-type {CONSTANT_SUBTYPE}:")
    for pathway in list(common_kegg)[:5]:
        print(f"      - {pathway}")

if spec1_kegg and len(spec1_kegg) > 0:
    print(f"\n   ✅ Voies spécifiques à {CONSTANT_SUBTYPE} vs {COMPARATOR1}:")
    for pathway in list(spec1_kegg)[:3]:
        print(f"      - {pathway}")

if spec2_kegg and len(spec2_kegg) > 0:
    print(f"\n   ✅ Voies spécifiques à {CONSTANT_SUBTYPE} vs {COMPARATOR2}:")
    for pathway in list(spec2_kegg)[:3]:
        print(f"      - {pathway}")

print("\n" + "="*60)
print("✅ ANALYSE TERMINÉE AVEC SUCCÈS !")
print("📁 Tous les graphiques ont été sauvegardés")
print("="*60)

# Sauvegarde des données transformées
df_transformed.to_csv('data_transformed_final.csv', index=False)
print("\n💾 Données transformées sauvegardées dans data_transformed_final.csv")
```

```python
# ============================================
# INTERPRÉTATION BIOLOGIQUE STRUCTURÉE - CORRIGÉE
# POUR LA PRÉSENTATION
# ============================================

print("\n" + "="*60)
print("🧬 INTERPRÉTATION BIOLOGIQUE - STRUCTURÉE POUR LA PRÉSENTATION")
print("="*60)

def generate_biological_interpretation(common, spec1, spec2, subtype, comp1, comp2):
    """
    Génère une interprétation biologique structurée
    """
    interpretation = f"""
╔══════════════════════════════════════════════════════════════════╗
║          INTERPRÉTATION BIOLOGIQUE DU SOUS-TYPE {subtype}          ║
╚══════════════════════════════════════════════════════════════════╝

1️⃣ CARACTÉRISTIQUES MOLÉCULAIRES DU SOUS-TYPE {subtype}
   ────────────────────────────────────────────────────────────────
"""

    # Vérification que common n'est pas None et a des éléments
    if common and len(common) > 0:
        interpretation += "\n   ✅ VOIES COMMUNES (signature du sous-type):\n"
        for i, term in enumerate(list(common)[:5]):
            term_lower = term.lower() if isinstance(term, str) else str(term).lower()
            # Classification
            if any(kw in term_lower for kw in ['cell cycle', 'division', 'mitosis', 'proliferation']):
                cat = "🔄 Prolifération cellulaire"
            elif any(kw in term_lower for kw in ['immune', 'inflammation', 'cytokine', 'interferon']):
                cat = "🛡️ Signalisation immunitaire"
            elif any(kw in term_lower for kw in ['dna repair', 'damage', 'replication']):
                cat = "🧬 Réparation de l'ADN"
            elif any(kw in term_lower for kw in ['hormone', 'estrogen', 'endocrine']):
                cat = "🧪 Signalisation hormonale"
            elif any(kw in term_lower for kw in ['metabolism', 'metabolic']):
                cat = "⚡ Métabolisme"
            elif any(kw in term_lower for kw in ['apoptosis', 'death', 'survival']):
                cat = "💀 Survie cellulaire"
            else:
                cat = "📌 Autre"
            interpretation += f"      • {term[:60]} → {cat}\n"
    else:
        interpretation += "\n   ⚠️ Aucune voie commune identifiée.\n"
        interpretation += "      → Cela peut indiquer que les deux comparaisons partagent\n"
        interpretation += "      des mécanismes moléculaires différents.\n"

    interpretation += f"""

2️⃣ COMPARAISON AVEC LE SOUS-TYPE {comp1}
   ────────────────────────────────────────────────────────────────
"""

    if spec1 and len(spec1) > 0:
        interpretation += "\n   🔴 VOIES SPÉCIFIQUES:\n"
        for term in list(spec1)[:3]:
            interpretation += f"      • {term[:60]}\n"
        interpretation += f"\n   ➡️ SIGNIFICATION: Ces voies montrent que le sous-type {subtype} se distingue de {comp1}\n"
        interpretation += f"      par des mécanismes moléculaires spécifiques qui peuvent expliquer\n"
        interpretation += f"      des différences de pronostic ou de réponse aux traitements.\n"
    else:
        interpretation += "\n   ⚠️ Aucune voie spécifique identifiée.\n"
        interpretation += f"      → Les profils moléculaires de {subtype} et {comp1} sont similaires\n"
        interpretation += f"      pour les voies analysées.\n"

    interpretation += f"""

3️⃣ COMPARAISON AVEC LE SOUS-TYPE {comp2}
   ────────────────────────────────────────────────────────────────
"""

    if spec2 and len(spec2) > 0:
        interpretation += "\n   🔵 VOIES SPÉCIFIQUES:\n"
        for term in list(spec2)[:3]:
            interpretation += f"      • {term[:60]}\n"
        interpretation += f"\n   ➡️ SIGNIFICATION: Ces voies reflètent les différences moléculaires entre {subtype} et {comp2},\n"
        interpretation += f"      suggérant des mécanismes de progression tumorale distincts.\n"
    else:
        interpretation += "\n   ⚠️ Aucune voie spécifique identifiée.\n"
        interpretation += f"      → Les profils moléculaires de {subtype} et {comp2} sont similaires\n"
        interpretation += f"      pour les voies analysées.\n"

    interpretation += f"""

4️⃣ IMPLICATIONS CLINIQUES
   ────────────────────────────────────────────────────────────────
   • Biomarqueurs potentiels : Les gènes et voies identifiés peuvent servir de marqueurs
     diagnostiques ou pronostiques pour le sous-type {subtype}.

   • Cibles thérapeutiques : Les voies spécifiques surexprimées représentent des cibles
     potentielles pour le développement de thérapies ciblées.

   • Médecine personnalisée : Ces résultats contribuent à une meilleure compréhension
     de l'hétérogénéité du cancer du sein et guident la stratigraphie des patients.

5️⃣ CONCLUSION
   ────────────────────────────────────────────────────────────────
   L'analyse transcriptomique a permis d'identifier des voies biologiques spécifiques
   au sous-type {subtype}, offrant des perspectives pour une meilleure compréhension
   des mécanismes moléculaires et pour le développement de stratégies thérapeutiques
   personnalisées.
"""

    return interpretation

# ============================================
# RÉCUPÉRATION DES DONNÉES POUR L'INTERPRÉTATION
# ============================================

print("\n🔍 Récupération des données pour l'interprétation...")

# Récupération des variables communes et spécifiques
# à partir des résultats d'enrichissement

def extract_pathway_sets(enrich_results_1, enrich_results_2, gene_set='KEGG_2021_Human'):
    """
    Extrait les ensembles de voies communes et spécifiques
    """
    # Initialisation avec des ensembles vides
    common = set()
    spec1 = set()
    spec2 = set()

    if gene_set in enrich_results_1 and gene_set in enrich_results_2:
        df1 = enrich_results_1[gene_set]
        df2 = enrich_results_2[gene_set]

        # Vérification que les DataFrames ne sont pas vides
        if df1 is not None and len(df1) > 0 and 'Term' in df1.columns:
            terms_1 = set(df1['Term'].tolist())
        else:
            terms_1 = set()
            print(f"   ⚠️ {gene_set} - Comparaison 1: Aucun terme trouvé")

        if df2 is not None and len(df2) > 0 and 'Term' in df2.columns:
            terms_2 = set(df2['Term'].tolist())
        else:
            terms_2 = set()
            print(f"   ⚠️ {gene_set} - Comparaison 2: Aucun terme trouvé")

        # Calcul des intersections
        common = terms_1 & terms_2 if terms_1 and terms_2 else set()
        spec1 = terms_1 - terms_2 if terms_1 else set()
        spec2 = terms_2 - terms_1 if terms_2 else set()

        print(f"   ✅ {gene_set}: {len(common)} communes, {len(spec1)} spécifiques 1, {len(spec2)} spécifiques 2")
    else:
        print(f"   ❌ {gene_set} non disponible dans les deux comparaisons")

    return common, spec1, spec2

# Récupération des données depuis les résultats d'enrichissement
# Si enrich_results_1 et enrich_results_2 existent
try:
    common_kegg, spec1_kegg, spec2_kegg = extract_pathway_sets(
        enrich_results_1, enrich_results_2, 'KEGG_2021_Human'
    )
except:
    print("   ⚠️ Erreur lors de l'extraction des données KEGG")
    common_kegg, spec1_kegg, spec2_kegg = set(), set(), set()

try:
    common_go, spec1_go, spec2_go = extract_pathway_sets(
        enrich_results_1, enrich_results_2, 'GO_Biological_Process_2021'
    )
except:
    print("   ⚠️ Erreur lors de l'extraction des données GO")
    common_go, spec1_go, spec2_go = set(), set(), set()

# ============================================
# GÉNÉRATION DE L'INTERPRÉTATION
# ============================================

print("\n" + "="*60)
print("📝 GÉNÉRATION DE L'INTERPRÉTATION")
print("="*60)

# Génération de l'interprétation pour KEGG
if common_kegg or spec1_kegg or spec2_kegg:
    interpretation_kegg = generate_biological_interpretation(
        common_kegg, spec1_kegg, spec2_kegg,
        CONSTANT_SUBTYPE, COMPARATOR1, COMPARATOR2
    )
    print("\n" + interpretation_kegg)

    # Sauvegarde pour la présentation
    with open('interpretation_biologique_KEGG.txt', 'w', encoding='utf-8') as f:
        f.write(interpretation_kegg)
    print("💾 Interprétation sauvegardée : interpretation_biologique_KEGG.txt")
else:
    print("\n⚠️ Aucune donnée KEGG disponible pour l'interprétation")
    print("   → Création d'une interprétation par défaut...")

    # Interprétation par défaut si pas de données
    default_interpretation = f"""
╔══════════════════════════════════════════════════════════════════╗
║          INTERPRÉTATION BIOLOGIQUE DU SOUS-TYPE {CONSTANT_SUBTYPE}          ║
╚══════════════════════════════════════════════════════════════════╝

📌 NOTE: Les données d'enrichissement n'ont pas pu être récupérées.
   Utilisez les données téléchargées depuis Enrichr pour compléter
   cette interprétation.

1️⃣ CARACTÉRISTIQUES MOLÉCULAIRES
   ────────────────────────────────────────────────────────────────
   Le sous-type {CONSTANT_SUBTYPE} est caractérisé par un profil d'expression
   distinct des sous-types {COMPARATOR1} et {COMPARATOR2}.

2️⃣ COMPARAISON AVEC {COMPARATOR1}
   ────────────────────────────────────────────────────────────────
   Les différences moléculaires entre {CONSTANT_SUBTYPE} et {COMPARATOR1}
   reflètent les mécanismes biologiques sous-jacents à ces sous-types.

3️⃣ COMPARAISON AVEC {COMPARATOR2}
   ────────────────────────────────────────────────────────────────
   Les différences moléculaires entre {CONSTANT_SUBTYPE} et {COMPARATOR2}
   suggèrent des voies de progression tumorale distinctes.

4️⃣ IMPLICATIONS CLINIQUES
   ────────────────────────────────────────────────────────────────
   Ces résultats contribuent à la compréhension de l'hétérogénéité
   du cancer du sein et guident le développement de thérapies ciblées.

5️⃣ CONCLUSION
   ────────────────────────────────────────────────────────────────
   L'analyse transcriptomique a permis d'identifier des signatures
   moléculaires spécifiques au sous-type {CONSTANT_SUBTYPE}.
"""
    print(default_interpretation)
    with open('interpretation_biologique_default.txt', 'w', encoding='utf-8') as f:
        f.write(default_interpretation)
    print("💾 Interprétation par défaut sauvegardée : interpretation_biologique_default.txt")

# ============================================
# GÉNÉRATION DE L'INTERPRÉTATION POUR GO (si disponible)
# ============================================

if common_go or spec1_go or spec2_go:
    interpretation_go = generate_biological_interpretation(
        common_go, spec1_go, spec2_go,
        CONSTANT_SUBTYPE, COMPARATOR1, COMPARATOR2
    )
    print("\n" + "="*60)
    print("🧬 INTERPRÉTATION - GO BIOLOGICAL PROCESS")
    print("="*60)
    print(interpretation_go)

    with open('interpretation_biologique_GO.txt', 'w', encoding='utf-8') as f:
        f.write(interpretation_go)
    print("💾 Interprétation GO sauvegardée : interpretation_biologique_GO.txt")

# ============================================
# TABLEAU RÉCAPITULATIF POUR LA PRÉSENTATION
# ============================================

print("\n" + "="*60)
print("📊 TABLEAU RÉCAPITULATIF POUR LA PRÉSENTATION")
print("="*60)

def create_summary_table(common, spec1, spec2, subtype, comp1, comp2, title="KEGG"):
    """
    Crée un tableau récapitulatif pour la présentation
    """
    print(f"\n📋 TABLEAU RÉCAPITULATIF - {title}")
    print("-"*80)
    print(f"{'Type':<30} {'Nombre de voies':<20} {'Exemples'}")
    print("-"*80)

    if common and len(common) > 0:
        examples = list(common)[:3]
        examples_str = ", ".join([ex[:30] + "..." if len(ex) > 30 else ex for ex in examples])
        print(f"{'🟣 Voies communes':<30} {len(common):<20} {examples_str}")
    else:
        print(f"{'🟣 Voies communes':<30} {'0':<20} {'Aucune'}")

    if spec1 and len(spec1) > 0:
        examples = list(spec1)[:3]
        examples_str = ", ".join([ex[:30] + "..." if len(ex) > 30 else ex for ex in examples])
        print(f"{'🔴 Spécifiques à ' + subtype + ' vs ' + comp1:<30} {len(spec1):<20} {examples_str}")
    else:
        print(f"{'🔴 Spécifiques à ' + subtype + ' vs ' + comp1:<30} {'0':<20} {'Aucune'}")

    if spec2 and len(spec2) > 0:
        examples = list(spec2)[:3]
        examples_str = ", ".join([ex[:30] + "..." if len(ex) > 30 else ex for ex in examples])
        print(f"{'🔵 Spécifiques à ' + subtype + ' vs ' + comp2:<30} {len(spec2):<20} {examples_str}")
    else:
        print(f"{'🔵 Spécifiques à ' + subtype + ' vs ' + comp2:<30} {'0':<20} {'Aucune'}")

    print("-"*80)

# Affichage des tableaux
create_summary_table(common_kegg, spec1_kegg, spec2_kegg,
                    CONSTANT_SUBTYPE, COMPARATOR1, COMPARATOR2, "KEGG")

create_summary_table(common_go, spec1_go, spec2_go,
                    CONSTANT_SUBTYPE, COMPARATOR1, COMPARATOR2, "GO Biological Process")

print("\n" + "="*60)
print("✅ INTERPRÉTATION TERMINÉE")
print("="*60)
```

```python
# ============================================
# RÉSEAU DE TERMES ENRICHIS AVEC CHEVAUCHEMENT GÉNIQUE
# ============================================

print("\n" + "="*60)
print("🌐 CONSTRUCTION DU RÉSEAU DE TERMES ENRICHIS")
print("="*60)

import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

def build_and_visualize_term_network(enrich_results_1, enrich_results_2,
                                      subtype_constant, comparator1, comparator2,
                                      gene_set='KEGG_2021_Human',
                                      overlap_threshold=2,
                                      top_n_terms=30):
    """
    Construit et visualise un réseau de termes enrichis avec chevauchement génique.

    Paramètres :
    -----------
    enrich_results_1 : dict - Résultats d'enrichissement pour la comparaison 1
    enrich_results_2 : dict - Résultats d'enrichissement pour la comparaison 2
    subtype_constant : str - Nom du sous-type constant
    comparator1 : str - Nom du premier comparateur
    comparator2 : str - Nom du deuxième comparateur
    gene_set : str - Nom de la base de données (ex: 'KEGG_2021_Human')
    overlap_threshold : int - Seuil minimal de chevauchement génique pour créer une arête
    top_n_terms : int - Nombre maximum de termes à inclure dans le réseau
    """

    # Vérification de la disponibilité des données
    if gene_set not in enrich_results_1 or gene_set not in enrich_results_2:
        print(f"⚠️ {gene_set} non disponible dans les résultats")
        return None

    df1 = enrich_results_1[gene_set]
    df2 = enrich_results_2[gene_set]

    # Vérification que les DataFrames ne sont pas vides
    if df1 is None or len(df1) == 0 or df2 is None or len(df2) == 0:
        print(f"⚠️ {gene_set}: Données insuffisantes")
        return None

    # Limitation du nombre de termes pour la lisibilité
    df1_top = df1.head(top_n_terms)
    df2_top = df2.head(top_n_terms)

    # Création du dictionnaire des gènes par terme
    term_genes = {}
    all_terms = set()

    for df in [df1_top, df2_top]:
        for _, row in df.iterrows():
            term = row['Term']
            all_terms.add(term)
            genes = row['Genes'] if isinstance(row['Genes'], list) else row['Genes'].split(';')
            if term not in term_genes:
                term_genes[term] = set()
            term_genes[term].update(genes)

    # Catégorisation des termes
    terms_1 = set(df1_top['Term'].tolist())
    terms_2 = set(df2_top['Term'].tolist())
    common_terms = terms_1 & terms_2
    specific_1 = terms_1 - terms_2
    specific_2 = terms_2 - terms_1

    print(f"\n📊 {gene_set}:")
    print(f"   - {subtype_constant} vs {comparator1}: {len(df1_top)} termes")
    print(f"   - {subtype_constant} vs {comparator2}: {len(df2_top)} termes")
    print(f"   - Termes communs: {len(common_terms)}")
    print(f"   - Spécifiques à {subtype_constant} vs {comparator1}: {len(specific_1)}")
    print(f"   - Spécifiques à {subtype_constant} vs {comparator2}: {len(specific_2)}")

    # Création du graphe
    G = nx.Graph()

    # Ajout des nœuds
    for term in all_terms:
        n_genes = len(term_genes[term])
        G.add_node(term,
                   label=term[:30],
                   n_genes=n_genes,
                   category='common' if term in common_terms else
                            'specific_1' if term in specific_1 else 'specific_2')

    # Ajout des arêtes (chevauchement génique)
    term_list = list(all_terms)
    edges_added = 0
    for i in range(len(term_list)):
        for j in range(i+1, len(term_list)):
            t1, t2 = term_list[i], term_list[j]
            overlap = len(term_genes[t1] & term_genes[t2])
            if overlap >= overlap_threshold:
                union = len(term_genes[t1] | term_genes[t2])
                jaccard = overlap / union if union > 0 else 0
                G.add_edge(t1, t2, weight=overlap, jaccard=jaccard)
                edges_added += 1

    print(f"\n🌐 Réseau construit:")
    print(f"   - {G.number_of_nodes()} nœuds (termes)")
    print(f"   - {G.number_of_edges()} arêtes (chevauchements ≥ {overlap_threshold} gènes)")

    # ============================================================
    # VISUALISATION DU RÉSEAU
    # ============================================================

    if len(G.nodes()) == 0:
        print("⚠️ Aucun nœud dans le réseau")
        return None

    plt.figure(figsize=(18, 14))

    # Positionnement des nœuds
    try:
        pos = nx.spring_layout(G, k=2.5, iterations=100, seed=42)
    except:
        pos = nx.spring_layout(G, k=2.5, iterations=100)

    # Couleurs par catégorie
    color_map = []
    for node in G.nodes():
        if node in common_terms:
            color_map.append('#9B59B6')  # Violet pour commun
        elif G.nodes[node]['category'] == 'specific_1':
            color_map.append('#E74C3C')  # Rouge pour spécifique 1
        else:
            color_map.append('#3498DB')  # Bleu pour spécifique 2

    # Taille des nœuds basée sur le nombre de gènes
    node_sizes = [G.nodes[node]['n_genes'] * 50 + 200 for node in G.nodes()]

    # Épaisseur des arêtes basée sur le chevauchement
    if G.number_of_edges() > 0:
        edge_weights = [G.edges[edge]['weight'] for edge in G.edges()]
        max_weight = max(edge_weights) if edge_weights else 1
        edge_widths = [w / max_weight * 3 + 0.5 for w in edge_weights]
    else:
        edge_widths = [0.5] * G.number_of_edges()

    # Dessin du réseau
    nx.draw(G, pos,
            node_color=color_map,
            node_size=node_sizes,
            with_labels=True,
            labels={node: node[:25] + '...' if len(node) > 25 else node for node in G.nodes()},
            font_size=8,
            font_weight='bold',
            edge_color='#BDC3C7',
            alpha=0.8,
            width=edge_widths if G.number_of_edges() > 0 else 0.5)

    # Légende personnalisée
    from matplotlib.patches import Patch
    legend_elements = [
        Patch(facecolor='#9B59B6', label=f'Commun aux deux comparaisons ({len(common_terms)})'),
        Patch(facecolor='#E74C3C', label=f'Spécifique à {subtype_constant} vs {comparator1} ({len(specific_1)})'),
        Patch(facecolor='#3498DB', label=f'Spécifique à {subtype_constant} vs {comparator2} ({len(specific_2)})')
    ]
    plt.legend(handles=legend_elements, loc='upper left', fontsize=12)

    plt.title(f'Réseau des termes enrichis ({gene_set})\n'
              f'Taille des nœuds = nombre de gènes | Épaisseur des arêtes = chevauchement génique\n'
              f'Seuil de chevauchement: ≥ {overlap_threshold} gènes | Top {top_n_terms} termes par comparaison',
              fontsize=14)
    plt.tight_layout()

    # Sauvegarde
    filename = f'term_network_{gene_set}.png'
    plt.savefig(filename, dpi=300, bbox_inches='tight')
    print(f"\n✅ Réseau sauvegardé : {filename}")
    plt.show()

    # ============================================================
    # AFFICHAGE DES TERMES COMMUNS ET SPÉCIFIQUES
    # ============================================================

    print("\n" + "-"*60)
    print("📋 LISTE DES TERMES")
    print("-"*60)

    if len(common_terms) > 0:
        print(f"\n🟣 Termes communs ({len(common_terms)}):")
        for i, term in enumerate(list(common_terms)[:10], 1):
            print(f"   {i}. {term}")
        if len(common_terms) > 10:
            print(f"   ... et {len(common_terms)-10} autres")

    if len(specific_1) > 0:
        print(f"\n🔴 Termes spécifiques à {subtype_constant} vs {comparator1} ({len(specific_1)}):")
        for i, term in enumerate(list(specific_1)[:10], 1):
            print(f"   {i}. {term}")
        if len(specific_1) > 10:
            print(f"   ... et {len(specific_1)-10} autres")

    if len(specific_2) > 0:
        print(f"\n🔵 Termes spécifiques à {subtype_constant} vs {comparator2} ({len(specific_2)}):")
        for i, term in enumerate(list(specific_2)[:10], 1):
            print(f"   {i}. {term}")
        if len(specific_2) > 10:
            print(f"   ... et {len(specific_2)-10} autres")

    return G, term_genes, common_terms


# ============================================================
# EXÉCUTION POUR DIFFÉRENTES BASES DE DONNÉES
# ============================================================

# Liste des bases de données à analyser
gene_sets_to_analyze = ['KEGG_2021_Human', 'GO_Biological_Process_2021', 'Reactome_2022']

print("\n" + "="*60)
print("🌐 CONSTRUCTION DES RÉSEAUX POUR PLUSIEURS BIBLIOTHÈQUES")
print("="*60)

# Dictionnaire pour stocker les résultats des réseaux
networks = {}

for gene_set in gene_sets_to_analyze:
    print(f"\n{'='*60}")
    print(f"📊 Analyse de {gene_set}")
    print("="*60)

    try:
        G, term_genes, common_terms = build_and_visualize_term_network(
            enrich_results_1,
            enrich_results_2,
            CONSTANT_SUBTYPE,
            COMPARATOR1,
            COMPARATOR2,
            gene_set=gene_set,
            overlap_threshold=2,
            top_n_terms=30
        )

        if G is not None:
            networks[gene_set] = {
                'graph': G,
                'term_genes': term_genes,
                'common_terms': common_terms
            }
    except Exception as e:
        print(f"❌ Erreur pour {gene_set}: {str(e)}")

print("\n" + "="*60)
print("✅ CONSTRUCTION DES RÉSEAUX TERMINÉE")
print("="*60)

# ============================================================
# TABLEAU RÉCAPITULATIF DES RÉSEAUX
# ============================================================

print("\n📊 TABLEAU RÉCAPITULATIF DES RÉSEAUX")
print("-"*60)

summary_data = []
for gene_set, data in networks.items():
    G = data['graph']
    common = data['common_terms']

    # Comptage des nœuds par catégorie
    n_common = len(common)
    n_specific_1 = sum(1 for n in G.nodes() if n not in common)
    n_specific_2 = sum(1 for n in G.nodes() if n not in common and G.nodes[n]['category'] == 'specific_2')

    summary_data.append({
        'Base de données': gene_set,
        'Nœuds (termes)': G.number_of_nodes(),
        'Arêtes': G.number_of_edges(),
        'Communs': n_common,
        'Spécifiques': G.number_of_nodes() - n_common
    })

summary_df = pd.DataFrame(summary_data)
print(summary_df.to_string(index=False))

# Sauvegarde du tableau
summary_df.to_csv('resume_reseaux_termes.csv', index=False)
print("\n💾 Tableau sauvegardé : resume_reseaux_termes.csv")

print("\n" + "="*60)
print("✅ ANALYSE DES RÉSEAUX TERMINÉE")
print("="*60)
```

```python
# ============================================
# TÉLÉCHARGEMENT DES IMAGES ET FICHIERS DU PROJET
# ============================================

print("\n" + "="*60)
print("📥 TÉLÉCHARGEMENT DES FICHIERS DU PROJET")
print("="*60)

import os
import shutil
from google.colab import files
import glob

# ============================================
# 1. LISTE DES FICHIERS À TÉLÉCHARGER
# ============================================

# Images du réseau de termes
network_images = [
    'term_network_KEGG_2021_Human.png',
    'term_network_GO_Biological_Process_2020.png',
    'term_network_Reactome_2022.png'
]

# Volcan plots
volcano_images = [
    'volcano_BRCA-Normal_vs_BRCA-LumA.png',
    'volcano_BRCA-Normal_vs_BRCA-Her2.png'
]

# Autres graphiques
other_images = [
    'heatmap_top_genes.png',
    'diagramme_flux.png'
]

# Fichiers d'enrichissement
enrichment_files = [
    'enrichment_BRCA-Normal_vs_BRCA-LumA_KEGG_2021_Human.png',
    'enrichment_BRCA-Normal_vs_BRCA-Her2_KEGG_2021_Human.png',
    'enrichment_BRCA-Normal_vs_BRCA-LumA_GO_Biological_Process_2020.png',
    'enrichment_BRCA-Normal_vs_BRCA-Her2_GO_Biological_Process_2020.png'
]

# Fichiers CSV
csv_files = [
    'DE_BRCA-Normal_vs_BRCA-LumA.csv',
    'DE_BRCA-Normal_vs_BRCA-Her2.csv',
    'resume_reseaux_termes.csv'
]

# Fichiers d'interprétation
interpretation_files = [
    'interpretation_biologique_KEGG.txt',
    'interpretation_biologique_GO.txt'
]

# ============================================
# 2. CRÉATION DU DOSSIER DE TÉLÉCHARGEMENT
# ============================================

download_dir = '/content/projet6_fichiers'
if not os.path.exists(download_dir):
    os.makedirs(download_dir)
print(f"✅ Dossier créé : {download_dir}")

# ============================================
# 3. COPIE DES FICHIERS
# ============================================

print("\n📋 Copie des fichiers...")

def copy_files(file_list, source_dir='/content', dest_dir=download_dir):
    """Copie une liste de fichiers vers le dossier de destination"""
    copied = 0
    not_found = []

    for filename in file_list:
        source_path = os.path.join(source_dir, filename)
        if os.path.exists(source_path):
            shutil.copy(source_path, dest_dir)
            print(f"   ✅ {filename}")
            copied += 1
        else:
            not_found.append(filename)

    return copied, not_found

# Copie des fichiers avec recherche alternative
def copy_files_with_alternative(file_list, source_dir='/content', dest_dir=download_dir):
    """Copie les fichiers en cherchant aussi dans les sous-dossiers"""
    copied = 0
    not_found = []

    for filename in file_list:
        found = False

        # Recherche dans le dossier source
        source_path = os.path.join(source_dir, filename)
        if os.path.exists(source_path):
            shutil.copy(source_path, dest_dir)
            print(f"   ✅ {filename}")
            copied += 1
            found = True

        # Recherche dans les sous-dossiers
        if not found:
            for root, dirs, filenames_in_dir in os.walk(source_dir):
                if filename in filenames_in_dir:
                    shutil.copy(os.path.join(root, filename), dest_dir)
                    print(f"   ✅ {filename} (trouvé dans {root})")
                    copied += 1
                    found = True
                    break

        if not found:
            not_found.append(filename)

    return copied, not_found

# Copie des images du réseau
print("\n🖼️ Images du réseau de termes :")
copy_files_with_alternative(network_images)

# Copie des volcan plots
print("\n📊 Volcan plots :")
copy_files_with_alternative(volcano_images)

# Copie des autres graphiques
print("\n📊 Autres graphiques :")
copy_files_with_alternative(other_images)

# Copie des fichiers d'enrichissement
print("\n📊 Graphiques d'enrichissement :")
copy_files_with_alternative(enrichment_files)

# Copie des fichiers CSV
print("\n📄 Fichiers CSV :")
copy_files_with_alternative(csv_files)

# Copie des fichiers d'interprétation
print("\n📝 Fichiers d'interprétation :")
copy_files_with_alternative(interpretation_files)

# ============================================
# 4. COPIE DU DOSSIER ENRICHR (si existe)
# ============================================

print("\n📁 Copie du dossier enrichr_results...")

enrichr_dir = '/content/enrichr_results'
if os.path.exists(enrichr_dir):
    enrichr_dest = os.path.join(download_dir, 'enrichr_results')
    if not os.path.exists(enrichr_dest):
        os.makedirs(enrichr_dest)

    for file in os.listdir(enrichr_dir):
        if file.endswith('.txt'):
            shutil.copy(os.path.join(enrichr_dir, file), enrichr_dest)
            print(f"   ✅ {file} copié")
else:
    print("   ⚠️ Dossier enrichr_results non trouvé")

# ============================================
# 5. COPIE DU NOTEBOOK (si existe)
# ============================================

print("\n📓 Copie du notebook...")

notebook_files = [f for f in os.listdir('/content') if f.endswith('.ipynb')]
if notebook_files:
    for nb in notebook_files:
        shutil.copy(f'/content/{nb}', f'{download_dir}/{nb}')
        print(f"   ✅ {nb} copié")
else:
    print("   ⚠️ Aucun notebook trouvé")

# ============================================
# 6. CRÉATION D'UN README POUR LE DOSSIER
# ============================================

print("\n📝 Création du README...")

readme_content = """
# Projet 6 - Fichiers de résultats

## 📊 Graphiques
- **term_network_*.png** : Réseau de termes enrichis (KEGG, GO, Reactome)
- **volcano_*.png** : Volcan plots (Normal vs LumA, Normal vs HER2)
- **heatmap_top_genes.png** : Heatmap des gènes différentiellement exprimés
- **diagramme_flux.png** : Diagramme de flux du pipeline
- **enrichment_*.png** : Barplots des voies enrichies

## 📄 Données
- **DE_*.csv** : Résultats de l'expression différentielle
- **resume_reseaux_termes.csv** : Résumé des réseaux de termes

## 📝 Interprétations
- **interpretation_biologique_*.txt** : Interprétations biologiques

## 📁 enrichr_results/
- Fichiers TXT des résultats Enrichr pour toutes les bases de données

## 📓 Notebook
- **.ipynb** : Notebook Colab complet du projet

## 🔗 Lien GitHub
https://github.com/jorthankia/projet6_breast_cancer
"""

with open(f'{download_dir}/README_fichiers.txt', 'w', encoding='utf-8') as f:
    f.write(readme_content)
print("   ✅ README_fichiers.txt créé")

# ============================================
# 7. AFFICHAGE DU CONTENU DU DOSSIER
# ============================================

print("\n" + "="*60)
print("📊 CONTENU DU DOSSIER DE TÉLÉCHARGEMENT")
print("="*60)

for root, dirs, filenames_in_dir in os.walk(download_dir):
    level = root.replace(download_dir, '').count(os.sep)
    indent = ' ' * 2 * level
    print(f'{indent}📁 {os.path.basename(root)}/')
    subindent = ' ' * 2 * (level + 1)
    for file in sorted(filenames_in_dir):
        file_size = os.path.getsize(os.path.join(root, file))
        if file_size > 1024 * 1024:
            size_str = f"{file_size/(1024*1024):.1f} Mo"
        elif file_size > 1024:
            size_str = f"{file_size/1024:.1f} Ko"
        else:
            size_str = f"{file_size} o"
        print(f'{subindent}📄 {file} ({size_str})')

# ============================================
# 8. COMPRESSION ET TÉLÉCHARGEMENT
# ============================================

print("\n" + "="*60)
print("📥 COMPRESSION ET TÉLÉCHARGEMENT")
print("="*60)

# Compression du dossier
zip_name = 'projet6_fichiers'
shutil.make_archive(zip_name, 'zip', download_dir)
print(f"✅ Dossier compressé : {zip_name}.zip")

# Téléchargement
files.download(f'/content/{zip_name}.zip')
print("\n📥 Téléchargement terminé !")

# ============================================
# 9. RÉSUMÉ FINAL
# ============================================

print("\n" + "="*60)
print("📋 RÉSUMÉ DES FICHIERS TÉLÉCHARGÉS")
print("="*60)

print(f"""
📁 {zip_name}.zip
│
├── 🖼️ IMAGES DU RÉSEAU (3)
│   ├── term_network_KEGG_2021_Human.png
│   ├── term_network_GO_Biological_Process_2021.png
│   └── term_network_Reactome_2022.png
│
├── 📊 VOLCAN PLOTS (2)
│   ├── volcano_BRCA-Normal_vs_BRCA-LumA.png
│   └── volcano_BRCA-Normal_vs_BRCA-Her2.png
│
├── 📊 AUTRES GRAPHIQUES
│   ├── heatmap_top_genes.png
│   ├── diagramme_flux.png
│   └── enrichment_*.png
│
├── 📄 DONNÉES CSV
│   ├── DE_BRCA-Normal_vs_BRCA-LumA.csv
│   ├── DE_BRCA-Normal_vs_BRCA-Her2.csv
│   └── resume_reseaux_termes.csv
│
├── 📝 INTERPRÉTATIONS
│   ├── interpretation_biologique_KEGG.txt
│   └── interpretation_biologique_GO.txt
│
├── 📁 enrichr_results/
│   └── *.txt
│
├── 📓 notebook_complet.ipynb
└── 📄 README_fichiers.txt
""")

print("="*60)
print("✅ TÉLÉCHARGEMENT TERMINÉ !")
print("="*60)
```
